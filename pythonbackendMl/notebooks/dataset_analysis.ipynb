```python
savingDir = Path("../dataset/data_wound_seg_256/train_images")
```

```python
for image_path in train_image:
    image = Image.open(image_path)
    resizedImage = image.resize(
        (256, 256),
        Image.Resampling.LANCZOS
    )
    resizedImage.save(savingDir/ image_path.name)
```

```python
len(list(Path(savingDir).glob("*.png")))
```

```python
savingMask = Path("../dataset/data_wound_seg_256/train_masks")
```
